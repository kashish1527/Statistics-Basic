#1.How do you calculate the mean, median, and mode of a dataset?

Mean: Add all numbers, divide by how many there are.
Example: (2 + 4 + 6) / 3 = 4

Median: Middle value when sorted.

Odd count → middle number

Even count → average of two middle numbers
Example: [1, 3, 5] → 3

Mode: Number that appears most often.
Example: [2, 2, 3, 4] → 2

```python
#2.Write a Python program to compute the variance and standard deviation of a dataset?

import math

# Sample dataset
data = [10, 12, 23, 23, 16, 23, 21, 16]

# Step 1: Calculate the mean
mean = sum(data) / len(data)

# Step 2: Calculate variance
variance = sum((x - mean) ** 2 for x in data) / len(data)

# Step 3: Calculate standard deviation
std_deviation = math.sqrt(variance)

# Print results
print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", std_deviation)
```

```python
#9.Write a Python script to calculate covariance between two datasets?

import numpy as np

# Sample datasets
x = [2, 4, 6, 8, 10]
y = [1, 3, 5, 7, 9]

# 1. Manual calculation of covariance
def calculate_covariance(x, y):
    if len(x) != len(y):
        raise ValueError("Datasets must be of equal length")
    mean_x = sum(x) / len(x)
    mean_y = sum(y) / len(y)
    cov = sum((xi - mean_x) * (yi - mean_y) for xi, yi in zip(x, y)) / len(x)
    return cov

# 2. Using NumPy
cov_matrix = np.cov(x, y, bias=True)  # bias=True for population covariance

# Output
print("Manual Covariance:", calculate_covariance(x, y))
print("NumPy Covariance Matrix:\n", cov_matrix)
print("NumPy Covariance (x vs y):", cov_matrix[0, 1])
```

```python
#10.Write a Python script to calculate the correlation coefficient between two datasets?

import numpy as np

# Sample datasets
x = [2, 4, 6, 8, 10]
y = [1, 3, 5, 7, 9]

# 1. Manual calculation of Pearson correlation coefficient
def calculate_correlation(x, y):
    if len(x) != len(y):
        raise ValueError("Datasets must be of equal length")
    mean_x = sum(x) / len(x)
    mean_y = sum(y) / len(y)

    numerator = sum((xi - mean_x) * (yi - mean_y) for xi, yi in zip(x, y))
    denominator = (sum((xi - mean_x) ** 2 for xi in x) * sum((yi - mean_y) ** 2 for yi in y)) ** 0.5

    return numerator / denominator

# 2. Using NumPy
correlation_matrix = np.corrcoef(x, y)  # Pearson correlation coefficient

# Output
print("Manual Correlation Coefficient:", calculate_correlation(x, y))
print("NumPy Correlation Coefficient Matrix:\n", correlation_matrix)
print("NumPy Correlation Coefficient (x vs y):", correlation_matrix[0, 1])
```

```python
#12.Implement and compare simple random sampling and systematic sampling?

import random
import numpy as np
import matplotlib.pyplot as plt

# Sample population (size 100)
population = list(range(1, 101))

# 1. Simple Random Sampling
def simple_random_sampling(population, sample_size):
    return random.sample(population, sample_size)

# 2. Systematic Sampling
def systematic_sampling(population, sample_size):
    step = len(population) // sample_size
    start = random.randint(0, step - 1)  # Random start point within the first step
    return [population[start + i * step] for i in range(sample_size)]

# Parameters
sample_size = 10

# Sampling
simple_random_sample = simple_random_sampling(population, sample_size)
systematic_sample = systematic_sampling(population, sample_size)

# Output
print("Simple Random Sample:", simple_random_sample)
print("Systematic Sample:", systematic_sample)

# Visual Comparison
plt.figure(figsize=(10, 5))

# Population distribution
plt.hist(population, bins=20, alpha=0.5, label='Population', color='gray', edgecolor='black')

# Simple Random Sampling - Scatter plot
plt.scatter(simple_random_sample, np.zeros_like(simple_random_sample), color='blue', label='Simple Random Sampling', zorder=5)

# Systematic Sampling - Scatter plot
plt.scatter(systematic_sample, np.zeros_like(systematic_sample), color='red', label='Systematic Sampling', zorder=5)

plt.title('Comparison of Simple Random Sampling and Systematic Sampling')
plt.xlabel('Population Values')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
#18.Implement Z-score normalization and explain its significance?

import numpy as np
import matplotlib.pyplot as plt

# Simulating a dataset with 1000 data points (normal distribution)
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)

# Calculate the mean and standard deviation of the dataset
mean = np.mean(data)
std_dev = np.std(data)

# Apply Z-score normalization
z_scores = (data - mean) / std_dev

# Output first few normalized values
print("First 5 Z-scores:", z_scores[:5])

# Plot original data and normalized data
plt.figure(figsize=(12, 6))

# Original data histogram
plt.subplot(1, 2, 1)
plt.hist(data, bins=30, color='lightblue', edgecolor='black')
plt.title("Original Data")
plt.xlabel("Value")
plt.ylabel("Frequency")

# Z-score normalized data histogram
plt.subplot(1, 2, 2)
plt.hist(z_scores, bins=30, color='lightgreen', edgecolor='black')
plt.title("Z-score Normalized Data")
plt.xlabel("Z-score Value")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
#20. Write a Python program to visualize covariance using a heatmap?

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Simulating two datasets (normal distribution)
np.random.seed(42)
data1 = np.random.normal(loc=50, scale=10, size=1000)
data2 = np.random.normal(loc=50, scale=15, size=1000)

# Create a DataFrame with both datasets
df = pd.DataFrame({'Dataset 1': data1, 'Dataset 2': data2})

# Calculate the covariance matrix
cov_matrix = df.cov()

# Plot the heatmap of the covariance matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cov_matrix, annot=True, cmap='coolwarm', fmt='.2f', cbar=True)

# Add titles and labels
plt.title("Covariance Heatmap")
plt.show()
```

```python
#22.Generate a dataset and implement both variance and standard deviation computations?

import numpy as np
import matplotlib.pyplot as plt

# Generating a dataset (normal distribution)
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)  # Mean = 50, Std = 10

# Manually calculate the variance
mean = np.mean(data)
variance_manual = np.sum((data - mean) ** 2) / len(data)

# Manually calculate the standard deviation
std_dev_manual = np.sqrt(variance_manual)

# Using NumPy to calculate the variance and standard deviation
variance_np = np.var(data)
std_dev_np = np.std(data)

# Output the results
print(f"Manual Variance: {variance_manual:.2f}")
print(f"Manual Standard Deviation: {std_dev_manual:.2f}")
print(f"NumPy Variance: {variance_np:.2f}")
print(f"NumPy Standard Deviation: {std_dev_np:.2f}")

# Plotting the data to visualize its distribution
plt.figure(figsize=(8, 6))
plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title("Generated Dataset Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
#24.Implement the Pearson and Spearman correlation coefficients for a dataset?

import numpy as np
import pandas as pd
from scipy.stats import pearsonr, spearmanr
import matplotlib.pyplot as plt
import seaborn as sns

# Generating two datasets
np.random.seed(42)
data1 = np.random.normal(loc=50, scale=10, size=1000)  # Normally distributed data
data2 = 0.5 * data1 + np.random.normal(loc=0, scale=5, size=1000)  # Linearly related with noise

# Create a DataFrame
df = pd.DataFrame({'Data1': data1, 'Data2': data2})

# Calculate Pearson correlation
pearson_corr, _ = pearsonr(df['Data1'], df['Data2'])

# Calculate Spearman correlation
spearman_corr, _ = spearmanr(df['Data1'], df['Data2'])

# Output the results
print(f"Pearson Correlation: {pearson_corr:.2f}")
print(f"Spearman Correlation: {spearman_corr:.2f}")

# Plotting the data and correlation visualization
plt.figure(figsize=(10, 6))

# Scatter plot to visualize the relationship between the datasets
sns.scatterplot(x=df['Data1'], y=df['Data2'], color='blue', alpha=0.6)
plt.title(f"Scatter Plot: Pearson={pearson_corr:.2f}, Spearman={spearman_corr:.2f}")
plt.xlabel("Data1")
plt.ylabel("Data2")
plt.show()
```
